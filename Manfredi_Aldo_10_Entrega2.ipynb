## 1. Carga Reproducible y Copia de Trabajo
Importamos las librerías necesarias, cargamos el conjunto de datos desde el repositorio oficial y creamos una copia explícita de trabajo para preservar el dataset original intacto... y la funcion paa la bitacora de registro ..

```python
import pandas as pd
import numpy as np

# Carga del dataset original con la URL correcta del repositorio
url = "https://raw.githubusercontent.com/manfredialdo/iad-tpi01/refs/heads/main/hotel%20booking%20TPI%20grupo%20J.csv"
df = pd.read_csv(url)

# Copia explícita de trabajo
df_limpio = df.copy()

# Inicialización de la Bitácora
registros_bitacora = []

def registrar(problema, variable, decision, justificacion):
    registros_bitacora.append({
        "problema_detectado": problema,
        "variable_afectada": variable,
        "decision_tomada": decision,
        "justificacion": justificacion
    })

print(f"Dataset original cargado correctamente: {df_limpio.shape[0]} filas y {df_limpio.shape[1]} columnas.")
```

```python
# A) Reservas sin huéspedes (adults + children + babies == 0)
# A) Identificación de duplicados por combinación de negocio (sin considerar booking_id)
subset_identificacion = [
    'hotel', 'arrival_date_year', 'arrival_date_month', 'arrival_date_day_of_month',
    'lead_time', 'adults', 'children', 'babies', 'adr', 'reserved_room_type'
]
duplicados_negocio = df_limpio.duplicated(subset=subset_identificacion, keep=False).sum()

registrar(
    problema=f"Se detectaron {duplicados_negocio} registros duplicados por combinación clave de reserva.",
    variable="Atributos combinados de reserva",
    decision="Conservar registros registrando el hallazgo.",
    justificacion="Pueden responder a reservas simultáneas de grupos gestionadas por la misma agencia."
)

# B) Conteo preliminar de reservas sin huéspedes
sin_huespedes = df_limpio[(df_limpio['adults'] + df_limpio['children'].fillna(0) + df_limpio['babies']) == 0]
cant_sin_huespedes = len(sin_huespedes)

if cant_sin_huespedes > 0:
    df_limpio = df_limpio.drop(sin_huespedes.index)
    registrar(
        problema=f"Existencia de {cant_sin_huespedes} reservas sin ningún huésped registrado.",
        variable="adults, children, babies",
        decision="Eliminar filas afectadas del dataset de trabajo.",
        justificacion="Una reserva requiere al menos 1 huésped para tener validez comercial."
    )

# C) Detección de ADR inconsistente (<= 0)
adr_invalidos = df_limpio[df_limpio['adr'] <= 0]
registrar(
    problema=f"Presencia de {len(adr_invalidos)} registros con ADR menor o igual a 0.",
    variable="adr",
    decision="Identificar valores no operativos para aislar en análisis tarifarios.",
    justificacion="Un valor de tarifa <= 0 distorsiona el cálculo del precio medio real."
)

# D) Outliers en adultos
adults_outliers = df_limpio[df_limpio['adults'] > 10]
if len(adults_outliers) > 0:
    registrar(
        problema=f"Valores atípicos en adultos (máximo de {df_limpio['adults'].max()}).",
        variable="adults",
        decision="Registrar inconsistencia de carga.",
        justificacion="Capacidad excede la ocupación estándar de una habitación individual."
    )
```

## 3. Diagnóstico de Calidad, Justificación y Tratamiento
Siguiendo la metodología exigida por la cátedra:
1. **Diagnóstico:** Se audita cuantitativamente el problema.
2. **Justificación:** Se analiza su impacto operativo e interpretativo en el negocio hotelero.

### A. Revisión de Duplicados Implícitos
* **Diagnóstico:** `df_limpio.duplicated().sum()` devuelve `0` debido a la columna `booking_id` (identificador único)[cite: 5]. Sin embargo, al evaluar la combinación de atributos clave de la reserva (`hotel`, fechas de llegada, `lead_time`, pasajeros, tarifa, tipo de habitación), se identifican **7,332 registros idénticos**[cite: 5].
* **Justificación:** No se deben eliminar automáticamente. En el sector hotelero, las agencias de viajes a menudo cargan lotes de reservas idénticas para diferentes personas de un mismo grupo. Suprimirlas reduciría erróneamente el volumen real de ventas.
* **Tratamiento:** Conservar los registros en el dataset e incluir la observación en la bitácora[cite: 5].

```python
# -------------------------------------------------------------------------
# 1. IDENTIFICACIÓN DE CLAVE ÚNICA Y DUPLICADOS IMPLÍCITOS
# -------------------------------------------------------------------------
# 1. Diagnóstico
subset_identificacion = [
    'hotel', 'arrival_date_year', 'arrival_date_month', 'arrival_date_day_of_month',
    'lead_time', 'adults', 'children', 'babies', 'adr', 'reserved_room_type'
]
duplicados_negocio = df_limpio.duplicated(subset=subset_identificacion, keep=False).sum()
print(f"Registros duplicados por combinación clave de negocio: {duplicados_negocio}")

# 2. Justificación y Bitácora
registrar(
    problema=f"Se detectaron {duplicados_negocio} registros idénticos por combinación clave de reserva.",
    variable="Atributos combinados de reserva",
    decision="Conservar registros registrando el hallazgo.",
    justificacion="Pueden responder a reservas simultáneas de grupos gestionadas en lote por la misma agencia."
)
```

### B. Inconsistencia Logica: Reservas con 0 Huéspedes
* **Diagnóstico:** Evaluamos las reservas donde la suma total de pasajeros (`adults + children + babies`) es igual a `0`[cite: 5].
* **Justificación:** Comercial y operacionalmente es imposible tener una estadía sin ocupantes registrados (*ghost booking* o fallo del sistema). Al no representar una operación real e impedir el cálculo de métricas por pasajero, su eliminación está plenamente justificada.
* **Tratamiento:** Identificar las **37 filas afectadas** y removerlas de `df_limpio`[cite: 5].

```python
# 1. Diagnóstico
sin_huespedes = df_limpio[(df_limpio['adults'] + df_limpio['children'].fillna(0) + df_limpio['babies']) == 0]
cant_sin_huespedes = len(sin_huespedes)
print(f"Cantidad de reservas registradas con 0 huéspedes: {cant_sin_huespedes}")

# 2. Justificación y Bitácora
if cant_sin_huespedes > 0:
    registrar(
        problema=f"Existencia de {cant_sin_huespedes} reservas sin ningún huésped registrado.",
        variable="adults, children, babies",
        decision="Eliminar filas afectadas del dataset de trabajo.",
        justificacion="Una reserva requiere al menos 1 huésped registrado para ser comercialmente válida."
    )
    # 3. Tratamiento
    df_limpio = df_limpio.drop(sin_huespedes.index)
```

### C. Tratamiento de Valores Faltantes (Nulos)
* **Diagnóstico:** Se detectan nulos en `agent` (3,479 / 13.92%), `company` (23,560 / 94.24%), `country` (117 / 0.47%) y `children`[cite: 5].
* **Justificación:**
  * `agent` y `company`: Los faltantes señalan que el cliente reservó de manera directa o no corporativa. Imputarlos con `0` mantiene el significado del negocio.
  * `country`: No se imputa con la moda (`PRT`) para no asentar el sesgo de que los datos faltantes sean locales. Se imputa con `'Unknown'`.
  * `children`: Representa ausencia de menores en el registro. Se imputa con `0`.
* **Tratamiento:** Aplicar imputaciones explícitas preservando la estructura del conjunto de datos[cite: 5].

```python
# 1. Diagnóstico de Nulos
print("Conteo inicial de nulos:")
print(df_limpio[['agent', 'company', 'country', 'children']].isna().sum())

# 2 y 3. Tratamiento y Registro en Bitácora

# Agent
df_limpio['agent'] = df_limpio['agent'].fillna(0).astype(int)
registrar("Faltantes en Agente", "agent", "Imputar con 0", "Valor nulo indica venta directa sin intermediación de agencia.")

# Company
df_limpio['company'] = df_limpio['company'].fillna(0).astype(int)
registrar("Faltantes en Empresa", "company", "Imputar con 0", "Valor nulo indica cliente particular no corporativo.")

# Country
df_limpio['country'] = df_limpio['country'].fillna('Unknown')
registrar("Faltantes en País", "country", "Imputar con 'Unknown'", "Evita sesgar la distribución de origen imputando la moda (PRT).")

# Children
df_limpio['children'] = df_limpio['children'].fillna(0).astype(int)
registrar("Faltantes en Niños", "children", "Imputar con 0", "Se asume ausencia de menores ante la falta de registro explícito.")
```

### D. Revisión de Tarifas e Inconsistencias Extremas
* **Diagnóstico:** Se identifican **415 registros** con tarifa diaria `adr <= 0` y valores extremos como reservas con más de 10 adultos[cite: 5].
* **Justificación:** Las tarifas `<= 0` responden a errores de facturación o promociones bonificadas que distorsionan el precio medio diario ($ADR$). Los picos de huéspedes pueden ser cargas grupales.
* **Tratamiento:** Registrar el hallazgo para aislar o filtrar estas observaciones en análisis de precios, conservando las filas en el dataset operativo para mantener el histórico de ocupación[cite: 5].

```python
# 1. Diagnóstico de Tarifas e Inconsistencias
adr_invalidos = df_limpio[df_limpio['adr'] <= 0]
print(f"Registros con ADR <= 0: {len(adr_invalidos)}")

# 2. Justificación y Bitácora
registrar(
    problema=f"Presencia de {len(adr_invalidos)} registros con ADR menor o igual a 0.",
    variable="adr",
    decision="Identificar valores no operativos para aislar en análisis tarifarios.",
    justificacion="Un valor de tarifa <= 0 distorsiona el cálculo del precio promedio real."
)

# Revisión de outliers en adultos
adults_outliers = df_limpio[df_limpio['adults'] > 10]
if len(adults_outliers) > 0:
    registrar(
        problema=f"Valores atípicos en adultos (máximo detectado: {df_limpio['adults'].max()}).",
        variable="adults",
        decision="Registrar inconsistencia de carga conservando el registro.",
        justificacion="Capacidad excede la ocupación estándar; requiere análisis separado para no distorsionar promedio por habitación."
    )
```

# 3. IMPUTACIÓN DE VALORES FALTANTES Y CADENAS

```python
df_limpio['agent'] = df_limpio['agent'].fillna(0).astype(int)
registrar("Faltantes en Agente", "agent", "Imputar con 0", "Valor nulo indica venta directa sin agencia.")

df_limpio['company'] = df_limpio['company'].fillna(0).astype(int)
registrar("Faltantes en Empresa", "company", "Imputar con 0", "Valor nulo indica reserva particular no corporativa.")

df_limpio['country'] = df_limpio['country'].fillna('Unknown')
registrar("Faltantes en País", "country", "Imputar con 'Unknown'", "Evita sesgar la distribución imputando la moda.")

df_limpio['children'] = df_limpio['children'].fillna(0).astype(int)
registrar("Faltantes en Niños", "children", "Imputar con 0", "Se asume ausencia de menores ante falta de registro.")
```

## 4. Transformación de Datos y Creación de Variables Derivadas
Con el objetivo de alimentar los análisis posteriores, construimos las 5 variables derivadas solicitadas por el TPI
1. `arrival_date_parsed`: Conversión formal de la fecha a formato `datetime`

2. `total_guests`: Suma de adultos, niños y bebés

3. `total_stay_nights`: Noches totales (fin de semana + días hábiles)

4. `is_family`: Indicador binario si la reserva incluye menores

5. `estimated_total_cost`: Ingreso total estimado ($ADR \times \text{noches}$)

6. `lead_time_group`: Discretización por tramos temporales de la antelación de reserva

```python
# Crear fecha de llegada unificada en formato datetime
# Parseo correcto usando la columna arrival_date existente (YYYY-MM-DD)
# Conversión formal de fecha
df_limpio['arrival_date_parsed'] = pd.to_datetime(df_limpio['arrival_date'], errors='coerce')

# Creación de variables requeridas
df_limpio['total_guests'] = df_limpio['adults'] + df_limpio['children'] + df_limpio['babies']
df_limpio['total_stay_nights'] = df_limpio['stays_in_weekend_nights'] + df_limpio['stays_in_week_nights']
df_limpio['is_family'] = ((df_limpio['children'] + df_limpio['babies']) > 0).astype(int)
df_limpio['estimated_total_cost'] = df_limpio['total_stay_nights'] * df_limpio['adr']

# Discretización de Lead Time
df_limpio['lead_time_group'] = pd.cut(
    df_limpio['lead_time'],
    bins=[-1, 7, 30, 90, 180, 999],
    labels=['Mismo día/Semana', 'Último mes', '1-3 Meses', '3-6 Meses', 'Más de 6 Meses']
)

print("Variables derivadas creadas y transformaciones completadas.")
```
