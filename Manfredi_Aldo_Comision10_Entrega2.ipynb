```python
import pandas as pd
import numpy as np

# 1. Carga reproducible y copia explícita
url = "https://raw.githubusercontent.com/manfredialdo/iad-tpi01/main/hotel%20booking%20TPI%20grupo%20J.csv"
df = pd.read_csv(url)
df_limpio = df.copy()

# Inicialización de Bitácora
registros_bitacora = []

def registrar(problema, variable, decision, justificacion):
    registros_bitacora.append({
        "problema_detectado": problema,
        "variable_afectada": variable,
        "decision_tomada": decision,
        "justificacion": justificacion
    })
```

```python
# Estructura de tipos de datos y conteo de nulos inicial
df_limpio.info()

# Resumen estadístico numérico
display(df_limpio.describe())

# Resumen estadístico categórico
display(df_limpio.describe(include=['object', 'category']))
```

# 2. Diagnóstico y Limpieza
## A. Duplicados Implícitos

```python
# 2. Diagnóstico y Limpieza de Calidad
# la columna booking_id hace que cada fila parezca única, así que filtramos por
# las variables operativas reales para detectar si existían reservas idénticas creadas en simultáneo por agencias de viajes.
# Definimos las columnas clave de una reserva para identificar y contar cuántos
#  registros son exactamente iguales entre sí (duplicados de negocio).
# A. Duplicados Implícitos
subset_identificacion = [
    'hotel', 'arrival_date_year', 'arrival_date_month', 'arrival_date_day_of_month',
    'lead_time', 'adults', 'children', 'babies', 'adr', 'reserved_room_type'
]
duplicados = df_limpio.duplicated(subset=subset_identificacion, keep=False).sum()
registrar(
    f"Se detectaron {duplicados} registros idénticos por combinación clave de reserva.",
    "Atributos combinados de reserva",
    "Conservar registros registrando el hallazgo.",
    "Pueden responder a reservas simultáneas de grupos gestionadas en lote por la misma agencia."
)

```

# 2. Diagnóstico y Limpieza
## C. supresion de nulos, conversion


```python
# C. supresión de Nulos
# reemplaza valores faltantes y cambia tipo de datos
df_limpio['agent'] = df_limpio['agent'].fillna(0).astype(int)
registrar("Faltantes en Agente", "agent", "Imputar con 0", "Valor nulo indica venta directa sin intermediación de agencia.")

df_limpio['company'] = df_limpio['company'].fillna(0).astype(int)
registrar("Faltantes en Empresa", "company", "Imputar con 0", "Valor nulo indica cliente particular no corporativo.")

df_limpio['country'] = df_limpio['country'].fillna("Unknown")
registrar("Faltantes en País", "country", "Imputar con 'Unknown'", "Evita sesgar la distribución de origen imputando la moda.")

df_limpio['children'] = df_limpio['children'].fillna(0).astype(int)
registrar("Faltantes en Niños", "children", "Imputar con 0", "Se asume ausencia de menores ante la falta de registro explícito.")
```

# diagnostico y limpieza
## Detección de Tarifas <= 0 y Outliers de Adultos

```python
adr_invalidos = df_limpio[df_limpio['adr'] <= 0]
adults_outliers = df_limpio[df_limpio['adults'] > 10]
```

```python
print(len(adr_invalidos), len(adults_outliers))
```

```python
registrar(
    f"Presencia de {len(adr_invalidos)} registros con ADR menor o igual a 0.",
    "adr",
    "Identificar valores no operativos para aislar en análisis tarifarios.",
    "Un valor de tarifa <= 0 distorsiona el cálculo del precio promedio real."
)
if len(adults_outliers) > 0:
    registrar(
        f"Valores atípicos en adultos (máximo detectado: {df_limpio['adults'].max()}).",
        "adults",
        "Registrar inconsistencia de carga conservando el registro.",
        "Capacidad excede la ocupación estándar; requiere análisis separado."
    )
```

# 3. Transformación de Datos y Variables Derivadas

```python
df_limpio['arrival_date_parsed'] = pd.to_datetime(df_limpio['arrival_date'], errors='coerce')
df_limpio['total_guests'] = df_limpio['adults'] + df_limpio['children'] + df_limpio['babies']
df_limpio['total_stay_nights'] = df_limpio['stays_in_weekend_nights'] + df_limpio['stays_in_week_nights']
df_limpio['is_family'] = ((df_limpio['children'] + df_limpio['babies']) > 0).astype(int)
df_limpio['estimated_total_cost'] = df_limpio['total_stay_nights'] * df_limpio['adr']

df_limpio['lead_time_group'] = pd.cut(
    df_limpio['lead_time'],
    bins=[-1, 7, 30, 90, 180, 999],
    labels=['Mismo día/Semana', 'Último mes', '1-3 Meses', '3-6 Meses', 'Más de 6 Meses']
)
```
